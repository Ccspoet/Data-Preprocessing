## Q10: Natural Farming Excellence
**Business Question:** What percentage of farming communities achieve high productivity (over 500 annual yield) without using fertilizers, demonstrating natural farming excellence?

**Context:** High-performing farms without fertilizer inputs represent efficient operations that could mentor others in sustainable practices.


```python
Natural_farming=Agri_tech[(Agri_tech['Annual_Yield'] >= 500) & (Agri_tech['Fertilizer_Type'].isna())]
Natural_farming
```

```python
Natural_farming['Community_Name']
```

```python
Agri_tech[(Agri_tech['Annual_Yield'] >= 500) & (Agri_tech['Fertilizer_Type'].isna())]
```

```python
All_Farming = Agri_tech.shape[0]
Natural_farming_count = Natural_farming.shape[0] 
Percentage_Farm = (Natural_farming_count/ All_Farming) * 100 

Percentage_Farm
```

### Comment
Nearly 1 in 5 high-performing farms (18.7%) use no recorded fertilizer. This proves that "Natural Farming" isn't just a niche or subsistence method—it is a scalable, high-output production model capable of competing directly with conventional chemical farming in this region.
