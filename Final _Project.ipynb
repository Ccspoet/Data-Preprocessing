## 📊 Dataset Description

You have been provided access to the council's mining database in a CSV file named `south_african_mining_data.csv` containing comprehensive operational data from mines across South Africa.

### Data Fields:

**Identification & Location:**
- **Mine_ID**: Unique mine identifier
- **Mine_Name**: Name of the mining operation
- **Province**: South African province (Gauteng, Northern Cape, Eastern Cape, etc.)
- **Primary_Mineral**: Main mineral extracted (Gold, Platinum, Diamonds, Coal, Manganese, Iron Ore, etc.)
- **Mine_Type**: Type of mining operation (Underground, Open-pit, Alluvial)
- **Ownership_Type**: Ownership structure (Private, State-Owned, Joint Venture)

**Operational Metrics:**
- **Annual_Production_Tons**: Annual production volume in metric tons
- **Employees**: Number of employees
- **Years_Operating**: How long the mine has been operational
- **Production_per_Employee**: Productivity metric (tons per employee)

**Safety Metrics:**
- **Safety_Incidents_2023**: Number of safety incidents in 2023
- **Safety_Rate_per_1000_Employees**: Safety incident rate normalized per 1000 employees

**Financial Metrics:**
- **Revenue_Million_ZAR**: Annual revenue in millions of South African Rand
- **Operating_Costs_Million_ZAR**: Annual operating costs in millions of ZAR
- **Profit_Million_ZAR**: Annual profit in millions of ZAR
- **Profit_Margin_Percent**: Profit margin percentage

**Sustainability Metrics:**
- **Environmental_Compliance_Score**: Environmental compliance score (0-100)
- **Community_Investment_Million_ZAR**: Investment in local communities in millions of ZAR
- **Water_Usage_Million_Liters**: Annual water consumption in millions of liters
- **Energy_Consumption_MWh**: Annual energy consumption in megawatt-hours

### Production Analysis

```python
# Production Volumes by Mineral
prod_by_mineral = mining_df.groupby('Primary_Mineral')['Annual_Production_Tons'].sum().sort_values(ascending=False)
print("Total Production by Mineral:\n", prod_by_mineral)
```

### Comment
Gold is the undisputed leader in production volume (~835,000 tons), almost double that of Diamonds. This aligns with South Africa's historical reputation as a gold-mining powerhouse.

There is a clear "Top 3" tier (Gold, Iron Ore, Diamonds) that significantly outproduces the rest. Copper and Chrome are the lowest by volume, suggesting they are either smaller sectors or harder/more expensive to extract in large quantities compared to the giants.

```python

# Plotting Production by Mineral
plt.figure(figsize=(10, 6))
sns.barplot(x=prod_by_mineral.index, y=prod_by_mineral.values, palette='viridis')
plt.title('Total Annual Production by Mineral')
plt.xticks(rotation=45)
plt.show()
```

### Comment
 Gold is the undisputed leader in production volume (~835,000 tons), producing nearly double the volume of Diamonds. This confirms that despite diversification efforts, the industry's output is still heavily anchored in gold mining.

 There is a distinct "upper tier" of commodities—Gold, Iron Ore, and Diamonds—which significantly outproduce the others. These three minerals likely drive the bulk of the sector's logistical requirements (rail and port capacity).

While South Africa is famous for precious metals (Gold, Platinum, Diamonds), the high volumes of Iron Ore (541k tons) and Coal (332k tons) show that industrial minerals are a critical, high-volume backbone of the mining economy, ensuring the sector isn't reliant solely on luxury commodity prices.

```python

# Productivity by Mine Type
productivity_by_type = mining_df.groupby('Mine_Type')['Production_per_Employee'].mean()
print("\nAverage Productivity by Mine Type:\n", productivity_by_type)
```

### Comment
Surprisingly, Underground mines are the most efficient in this dataset, with an average of ~26.4 tons per employee. This is nearly double the productivity of Alluvial mines (~12.1) and significantly higher than Open Pit mines (~14.9).

This result contradicts the general industry assumption that Open Pit mining (with its massive haul trucks and surface access) is always more productive per capita. The data suggests that the Underground mines in this portfolio are likely highly mechanized or focused on high-volume bulk commodities (like Coal), whereas the Open Pit operations might be smaller or more labor-intensive than expected.

```python

# High and Underperforming Mines
top_performers = mining_df.nlargest(5, 'Production_per_Employee')[['Mine_Name', 'Production_per_Employee']]
low_performers = mining_df.nsmallest(5, 'Production_per_Employee')[['Mine_Name', 'Production_per_Employee']]

print("\nTop 5 Performing Mines:\n", top_performers)
print("\nBottom 5 Performing Mines:\n", low_performers)
```

### Comment
Mine_215 is a statistical anomaly. With 974.9 tons per employee, it is nearly double the second-best mine and almost 50 times higher than the industry average (~20 tons). This suggests Mine_215 is likely a highly automated, open-pit operation extracting a bulk commodity (like Coal or Iron Ore) where volume is easy to move with few people.

 The bottom 5 mines have productivity rates of 0.0 to 0.2 tons. This is not just "low efficiency"; it is effectively zero production. These mines are likely not operational. They are probably in "Care and Maintenance" mode, where staff are employed solely to secure the site and maintain machinery, but no actual mining is taking place. They are cost centers, not revenue generators.

```python
# Relationship between Mine Age and Efficiency
plt.figure(figsize=(10, 6))
sns.scatterplot(data=mining_df, x='Years_Operating', y='Production_per_Employee', hue='Mine_Type')
plt.title('Mine Age vs. Production Efficiency')
plt.show()
```

### Comment
 The plot reveals that the dominance of Gold and Iron Ore is driven by a few massive "mega-mines" (dots situated high on the y-axis). The majority of dots for these minerals are actually clustered lower down, proving that the sector relies heavily on these few giants rather than broad-based high production.

In contrast, minerals like Copper and Chrome show points tightly clustered near the bottom with no extreme vertical outliers. This indicates these sectors are composed almost entirely of small-to-medium operations, lacking the massive industrial scale seen in the Gold sector.

For commodities like Coal, there is a visible gap between the dense cluster of small mines at the bottom and the few high-performers at the top. This suggests a polarized industry structure: mines are either small-scale/boutique or massive conglomerates, with very few mid-sized operators.

```python

# Correlation check
correlation = mining_df['Years_Operating'].corr(mining_df['Production_per_Employee'])
print(f"Correlation between Age and Efficiency: {correlation}")
```

### Comment
 The correlation is effectively zero (-0.038). This is a critical finding because it defies the common assumption that older mines become less efficient due to deeper reserves or outdated infrastructure.

 In this dataset, a 50-year-old mine is just as likely to be highly productive as a 2-year-old mine. This suggests that operational factors (like management, automation, or mineral grade) matter far more than the physical age of the site. Experience does not guarantee efficiency, but neither does youth.

### Safety Performance Analysis

```python
# Safety Incidents by Province
safety_by_province = mining_df.groupby('Province')['Safety_Incidents_2023'].sum().sort_values(ascending=False)
print("Safety Incidents by Province:\n", safety_by_province)
```

### Comment
The Western Cape records the highest number of incidents (135). This is consistent with earlier findings that the Western Cape is the top producer by volume. It suggests a correlation where higher operational activity leads to a higher frequency of incidents, simply due to the scale of work being done.

 Gauteng follows closely with 124 incidents. Given that Gauteng mines are often deep-level gold operations (historically more hazardous), this high number likely reflects the inherent dangers of the specific mining methods used in this province compared to open-pit regions.

There is a significant gap between the most dangerous province (Western Cape, 135) and the least (Limpopo, 72). This nearly 2x difference indicates that safety challenges are not uniform across the country and national safety policies might need to be tailored to specific provincial risk profiles.

```python

# Safety Rate by Mineral Sector
safety_by_mineral = mining_df.groupby('Primary_Mineral')['Safety_Rate_per_1000_Employees'].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(x=safety_by_mineral.index, y=safety_by_mineral.values, palette='Reds_r')
plt.title('Average Safety Rate per 1000 Employees by Mineral')
plt.xticks(rotation=45)
plt.show()
```

### COmment
 Manganese (~6.8) and Iron Ore (~6.4) have the highest safety incident rates. This is counter-intuitive, as these are often Open Pit mines assumed to be safer. It suggests that the massive scale of material movement (heavy machinery, haul trucks) in these sectors creates significant hazards, distinct from the rock-fall risks of deep mining.

 Surprisingly, Gold (~5.1) and Platinum (~5.0) have the lowest safety rates. Historically considered the most dangerous due to depth, their low rates likely reflect mature, stringent safety protocols and heavy regulation that have successfully mitigated risks over decades.

 The difference between the "most dangerous" (Manganese) and "safest" (Platinum) sectors is relatively small (~1.8 points). This indicates that safety risks are a systemic industry-wide challenge rather than being isolated to a single "rogue" sector. Interventions should focus on universal hazards like machinery operation and fatigue management.

```python

# Relationship between Mine Size (Employees) and Safety Rate
plt.figure(figsize=(10, 6))
sns.scatterplot(data=mining_df, x='Employees', y='Safety_Rate_per_1000_Employees', hue='Mine_Type')
plt.title('Mine Size (Employees) vs. Safety Rate')
plt.show()
```

### Comment
There is a visible negative correlation (downward trend). As the number of employees increases (moving right), the safety rate generally decreases (moving down).

 The highest safety rates (outliers above 12.5) are almost exclusively found in smaller mines (fewer than 480 employees). This suggests that smaller operations may lack the resources, dedicated safety officers, or training infrastructure of their larger counterparts.

 Mines with over 540 employees show much tighter clustering with lower safety rates, indicating that scale brings stability and better standardized safety procedures.

### Economic Performance Analysis**

```python
# Profitability by Mineral
profit_by_mineral = mining_df.groupby('Primary_Mineral')['Profit_Margin_Percent'].mean().sort_values(ascending=False)
print("Average Profit Margin by Mineral:\n", profit_by_mineral)
```

### Comment

The most striking finding is that every single mineral sector is operating at a loss on average. There is no positive profit margin in the entire industry dataset. This confirms that the "Economic Pressure" mentioned in the business context is a severe, industry-wide crisis, not just a fluctuation.

 Manganese (-6.99%) and Copper (-8.08%) are the most resilient sectors. While still unprofitable, they are closest to the break-even point ($0 profit), suggesting they have better cost controls or more favorable market prices than the rest of the industry.

 Despite being the leader in production volume (as seen in the production analysis), Gold is the least profitable sector with a catastrophic margin of -104.62%.


```python

# Revenue vs Costs
plt.figure(figsize=(10, 6))
sns.scatterplot(data=mining_df, x='Operating_Costs_Million_ZAR', y='Revenue_Million_ZAR', hue='Ownership_Type')
plt.plot([0, mining_df['Operating_Costs_Million_ZAR'].max()], [0, mining_df['Operating_Costs_Million_ZAR'].max()], color='red', linestyle='--') # Break-even line
plt.title('Revenue vs. Operating Costs')
plt.show()
```

### Comment
The most critical observation is that the vast majority of data points fall below the red dashed line. This visually confirms that most mines in South Africa are spending more on operations than they are earning in revenue. The density of points below the line is alarming and points to a systemic profitability crisis.

The few mines that are significantly above the red line (highly profitable) are almost exclusively Private (blue dots). This suggests that private ownership structures in this dataset are better at managing costs or maximizing revenue than Joint Ventures or State-Owned entities.

```python

# Identifying Struggling Operations
struggling_mines = mining_df[mining_df['Profit_Million_ZAR'] < 0]
print(f"Number of Unprofitable Mines: {len(struggling_mines)}")
print(struggling_mines[['Mine_Name', 'Profit_Million_ZAR']].head())
```

### Comment
The analysis identifies 123 unprofitable mines out of the total 300 in the dataset. This means 41% of the entire mining sector is operating at a loss. This is a staggering figure that confirms the industry is facing a widespread solvency crisis, likely driven by the "rising operational costs" and "fluctuating commodity prices" mentioned in the business context.

The losses are not trivial. Mine_3 alone lost 88.91 Million ZAR in a single year, and Mine_7 lost 65.22 Million ZAR. These are massive sums of capital destruction.

 With over 40% of mines in the red, there is a high risk of imminent mine closures. This poses a significant threat not just to investors, but to the "hundreds of thousands of workers" mentioned in the introduction who rely on these specific mines for employment.

```python

#  Impact of Efficiency on Profitability
plt.figure(figsize=(10, 6))
sns.regplot(data=mining_df, x='Production_per_Employee', y='Profit_Margin_Percent')
plt.title('Operational Efficiency vs. Profit Margin')
plt.show()
```

### Comment
The regression line is almost flat (with a very slight positive slope). This indicates a near-zero correlation between operational efficiency and profitability.

 The dense cluster of points on the far left (low efficiency) shows the wildest variation in losses (ranging from -50% down to -2000%). This confirms that while high efficiency doesn't guarantee profit, low efficiency leaves mines completely exposed to massive financial losses.

### Sustainability Assessment

```python
# Environmental Compliance
avg_compliance = mining_df['Environmental_Compliance_Score'].mean()
print(f"Industry Average Environmental Compliance Score: {avg_compliance:.2f}")

plt.figure(figsize=(8, 5))
sns.histplot(mining_df['Environmental_Compliance_Score'], bins=20, kde=True)
plt.title('Distribution of Environmental Compliance Scores')
plt.show()
```

### Comment
The industry average is 80.65 out of 100. This is a surprisingly strong result given the financial distress in the sector. It suggests that despite cutting costs, mines remain committed to environmental standards—likely due to stringent South African mining regulations and the risk of license suspension for non-compliance.

 The most important observation is the absence of low scores. The x-axis starts at 60, meaning there are virtually no mines in this dataset with a score below 60. This suggests a strict regulatory environment where falling below a certain threshold likely results in immediate closure, forcing a high baseline of compliance.

The first peak is around 80, representing the "standard" operating procedure for the majority of mines.
The second, sharp peak is between 90 and 95, indicating a significant group of "Sustainability Leaders" who are exceeding basic requirements

```python

# Community Investment by Ownership
community_invest = mining_df.groupby('Ownership_Type')['Community_Investment_Million_ZAR'].mean()
print("\nAverage Community Investment by Ownership:\n", community_invest)
```

### Comment
Investment levels are virtually identical across all three groups (~2.0 to 2.1 million ZAR). There is no "generous" ownership type.

Contrary to expectation, State-Owned mines (2.09M) do not invest significantly more than Private mines (2.01M).

The lack of variation suggests this spending is driven by strict regulatory mandates (like Social and Labour Plans) rather than voluntary corporate initiatives. Everyone is paying exactly what they are forced to pay.

```python

# Resource Efficiency (Water per Ton)
mining_df['Water_Efficiency'] = mining_df['Water_Usage_Million_Liters'] / mining_df['Annual_Production_Tons']

# Best practices (Lowest water usage per ton)
water_efficient_mines = mining_df.nsmallest(5, 'Water_Efficiency')[['Mine_Name', 'Water_Efficiency', 'Primary_Mineral']]
print("\nTop 5 Water Efficient Mines:\n", water_efficient_mines)
```

### Comment

 Mine_215 is the first water-efficient mine (0.0016), just as it was the #1 labor-efficient mine. This proves that operational excellence is holistic; mines that optimize for workforce productivity are also optimizing for resource usage. It is likely a modern, fully integrated "Smart Mine."

 The top 5 includes Gold, Iron Ore, Manganese, and Diamonds. This demonstrates that high water efficiency is achievable across all commodity types, regardless of the extraction method (wet vs. dry processing).

 The difference between the 1st and 5th mine is extremely small. This suggests the top performers have hit a technology ceiling, likely all utilizing similar state-of-the-art water recycling or dry-stack tailings methods.

### Regional and Mineral Type Comparisons

### Comment
The top three performing provinces are Eastern Cape, KwaZulu-Natal, and Western Cape. All three are coastal regions. This implies a massive logistical advantage; mines in these provinces likely have significantly lower transport costs to ports compared to their inland counterparts, directly boosting their bottom line.

 Gauteng is the only province showing a net loss (a negative bar below the zero line). Since Gauteng is historically the center of gold mining, this visually confirms the earlier data that the gold sector is financially insolvent. The traditional heart of the industry is now its biggest financial liability.

```python

# Energy Consumption by Mineral
energy_by_mineral = mining_df.groupby('Primary_Mineral')['Energy_Consumption_MWh'].mean().sort_values(ascending=False)
print("Average Energy Consumption by Mineral:\n", energy_by_mineral)
```

### Comment
Iron Ore is significantly more energy-intensive than any other sector, averaging ~2,914 MWh. This is nearly 1,000 MWh higher than the second-highest consumer (Coal). This huge gap suggests that the beneficiation process for Iron Ore (crushing, screening, and heavy media separation) requires vastly more power than other mineral processing methods.

Coal mines are the second highest consumers (~1,976 MWh). This creates a negative feedback loop where the extraction of energy (coal) itself consumes massive amounts of energy, likely cutting into the "net energy" return of these mines.

```python

#  Underground vs Open-pit Compliance
compliance_by_type = mining_df.groupby('Mine_Type')['Environmental_Compliance_Score'].mean()
print("\nCompliance Score by Mine Type:\n", compliance_by_type)
```

 ### Comment
 Alluvial mines are the clear leaders in sustainability, with an average score of ~83.9. This makes sense physically: alluvial mining (often in riverbeds or surface deposits) typically requires less heavy blasting, crushing, and harsh chemical processing than hard-rock mining, making it easier to meet environmental targets.

 Underground (~80.6) and Open Pit (~80.1) scores are virtually identical. This suggests that the heavy industrial nature of these methods caps their environmental performance at a specific "ceiling," regardless of whether the operation digs deep shafts or massive surface pits.

 The fact that even the lowest-performing category averages over 80 reinforces the earlier finding: strict environmental compliance is a non-negotiable baseline across the entire sector. No specific mining method is being allowed to cut corners.
